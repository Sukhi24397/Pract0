```python
#1.Tokenization
import nltk
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('punkt_tab')
```

```python
from nltk.sem.drt import Tokens
from nltk.tokenize  import word_tokenize
text= "Natural language processing is interesting"
tokens=word_tokenize(text)
print(tokens)
```

```python
#sentence tokanization
from nltk.tokenize import sent_tokenize

text = "Python is easy.It is widely used in data science"
sentencess = sent_tokenize(text)
print(sentencess)
```

```python
#stop words removal
#stop words such as is,the,and,in are removed

from nltk.corpus import stopwords
stop_words = set(stopwords.words('english'))

words = word_tokenize(text)
filtered_words = [word for word in words if word.casefold() not in stop_words]
print(filtered_words)
```

```python
#steming
#steming reduces words to their root form
from nltk.stem import PorterStemmer
stemmer = PorterStemmer()
words = ["playing","studies","running","gone","watched","coding"]
for word in words:
    print(stemmer.stem(word))
```

```python
#lematization
#lematization converts words to their dictionary forms
from nltk.stem import WordNetLemmatizer
lemmatizer = WordNetLemmatizer()
print(lemmatizer.lemmatize("writing"))
print(lemmatizer.lemmatize("studies"))
print(lemmatizer.lemmatize("best"))
print(lemmatizer.lemmatize("running"))
print(lemmatizer.lemmatize("coding"))
```

```python
#part of speech(POS)tagging
#POS tagging identifies gramatical roles of words
import nltk
nltk.download('averaged_perceptron_tagger')
nltk.download('averaged-perceptron_tagger_end')
```

```python
import nltk
nltk.download('averaged_perceptron_tagger_eng')
from nltk import pos_tag
from nltk.tokenize import word_tokenize

words = word_tokenize("python is a powerful programming language")
pos_tags = pos_tag(words)
print(pos_tags)
```

```python
#word frequency count
from nltk.probability import FreqDist
text = "Natural language processing is interesting"
words = word_tokenize(text)
fd = FreqDist(words)
print(fd.most_common())
```
