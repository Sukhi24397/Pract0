```python
import pandas as pd
import numpy as np
data = {
    'Product':['Laptop','Mobile','Tablet','Headphones','Smartwatch','Camera'],
    'Price':[55000,20000,30000,2000,8000,45000],
    'Units_Sold':[25,60,35,80,50,15]
}
df= pd.DataFrame(data)
print(df)
```

```python
#Data Binning
bins1 = [0,10000,30000,60000]
labels1 = ['Low','Medium','High']
df['Price_Category'] = pd.cut(df['Price'],bins=bins1,labels=labels1)
df.head()
```

```python
#binning uits sold
bins2 = [0,30,60,100]
labels2 = ['Low Sales','Medium Sales','High Sales']
df['Sales_Category'] = pd.cut(df['Units_Sold'],bins=bins2,labels=labels2)
print(df)
```

```python
#data formatting
#convert data types

print(df.dtypes)
df['Units_Sold']=df['Units_Sold'].astype(float)
print(df.dtypes)
df
```

```python
#convert text to uppercse
df['Product']=df['Product'].str.upper()
print(df)
```

```python
#convert text to lowercse
df['Product']=df['Product'].str.lower()
print(df)
```

```python
#rounding values
df['Price'] = df['Price'].round(2)
print(df)
```

```python
#additional operations
#sort data
sorted_df = df.sort_values(by='Price')
print(sorted_df)
```

```python
df.sort_values(by='Units_Sold',inplace=True)
print(df)
```

```python
df.sort_values(by='Units_Sold',ascending=False,inplace=True)
print(df)
```

```python
#display unique values
print(df['Price_Category'].unique())
```

```python
#dataset:food delivery orders
import pandas as pd
import numpy as np
data = {
'Order_Id': [1,2,3,4,5,6,7,8],
'Order_Value' :[250,450,120,600,300,150,500,220],
'Delivery_Time' :[30,45,25,60,35,20,50,28],
'Distance_km' :[5,8,3,10,6,2,9,4]
}
df = pd.DataFrame(data)
print(df)
```

```python
#binning delivery time
bins2 = [0,30,45,70]
labels2 = ['Fast','Medium','Slow']
df['Delivery_Speed'] = pd.cut(df['Delivery_Time'],bins=bins2,labels=labels2)
print(df)
```

```python
# Binning Distance_km

bins_distance = [0, 4, 8, 12]

labels_distance = ["Short Distance", "Medium Distance", "Long Distance"]

df['Distance_Category'] = pd.cut(df['Distance_km'], bins=bins_distance, labels=labels_distance)

df
```

```python
#count of categories
# Binning Order_Value
bins_order_value = [0, 200, 400, 600, 800]
labels_order_value = ['Low', 'Medium', 'High', 'Very High']
df['Order_Category'] = pd.cut(df['Order_Value'], bins=bins_order_value, labels=labels_order_value, include_lowest=True)
print(df['Order_Category'].value_counts())
print(df)
```
