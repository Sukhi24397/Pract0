```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np
```

```python
import numpy as np
import pandas as pd

np.random.seed(0)
df= pd.DataFrame({
    'Category':['A','B','C','D','E'],
    'Vales':[0,70,30,90,60],
    'Sales':np.random.randint(1,100,size=5),
    'Profit':np.random.randint(10,100,5)

})
print(df)
```

```python
import matplotlib.pyplot as plt
#1. Area Plot
plt.figure()
plt.fill_between(df['Category'], df['Vales'], color='green', alpha=0.25)
plt.title('Sukhi')
plt.xlabel('Category')
plt.ylabel('Values')
plt.show()
print(df)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_style("whitegrid") #apply seaborn style
plt.figure(figsize=(7,4))
plt.fill_between(df['Category'], df['Sales'],
                 color = 'skyblue', alpha = 0.5, label = 'Sales')
plt.fill_between(df['Category'],df['Profit'],
                 color = 'orange', alpha = 0.5, label = 'Profit')
plt.title("Sukhi")
plt.xlabel("Category")
plt.ylabel("Values")
plt.legend()
print(df)
```

```python
#2. Pie Chart
plt.figure()
plt.pie(df['Vales'], labels=df['Category'], autopct = '%1.2f%%')
plt.title('Sukhi')
plt.show()
```

```python
#3. Donut Chart
plt.figure()
plt.pie(df['Vales'], labels=df['Category'], autopct='%1.1f%%')
centre_circle = plt.Circle ((0,0), 0.70, fc='white')
fig = plt.gcf()
fig.gca().add_artist(centre_circle)
plt.title("Donut Chart")
plt.show()
```

```python
#4. Boxplot (Outliners Detection)
plt.figure()
sns.boxplot(x=df['Vales'])
plt.title('Sukhi')
plt.show()
```

```python
#5. Heatmap (Correlation)
plt.figure()
corr = df[['Sales', 'Profit']].corr()
sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.title('Sukhi')
plt.show
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
# Select only numeric columns
corr = df[['Vales', 'Sales', 'Profit']].corr()
plt.figure(figsize=(5,4))
sns.heatmap(corr, annot=True)
plt.title("Correlation Heatmap")
```

```python
#Bubble Plot (3 Variable Chart)
import matplotlib.pyplot as plt
plt.scatter(df['Sales'], df['Profit'], s=df['Vales']*10) #Size of Bubbles
plt.xlabel('Sales')
plt.ylabel('Profit')
plt.title('Sukhi')
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.scatterplot(x='Sales', y='Profit',
                size='Vales', #Bubble Size
                hue='Vales',  #Colour
                data=df,
                sizes=(50,300), #Min-Max Size
                palette = 'viridis')
plt.title("Sukhi")
plt.show()
```
